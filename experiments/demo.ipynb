```python
import numpy as np
import scipy
from scipy.special import entr
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5, 5, 1000)

f = lambda x: (np.sqrt(1 + x ** 2) - x) / 2
g = lambda x: (np.sqrt(1 + x ** 2) + x) / 2
y = 1 + f(x)
z = (1 + g(x)) * np.exp(entr(y))
r = (1 + g(x)) * np.exp(- f(x) ** 2)
s = np.log(1 + np.exp(x))
plt.semilogy(x, z, x, r, x, s)
```
